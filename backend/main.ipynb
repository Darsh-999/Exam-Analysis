# Classification (question -> topic/subtopic mapping)

```python
import asyncio

from app.classification_service.postprocess import resolve_question_mappings
from app.classification_service.preprocess import assign_question_ids, assign_topic_ids, chunked
from app.config import settings
from app.local_classification_service.classify import classify_batch
```

```python
# assign_topic_ids/assign_question_ids normally read Mongo docs (a syllabus
# doc's `extracted_data` field, a `questions` collection cursor). There's no
# DB in this notebook, so we fake that shape from the responses already
# extracted above -- one syllabus, with question `_id`s standing in for what
# would otherwise be Mongo ObjectIds.
syllabi_docs = [{"extracted_data": topic_response}]
question_docs = [
    {"_id": i, "question": q["question"]}
    for i, q in enumerate(paper_response["content"])
]

topics_and_subtopics, topic_lookup, subtopic_lookup = assign_topic_ids(syllabi_docs)
prompt_questions, id_map = assign_question_ids(question_docs)
batches = chunked(prompt_questions, settings.classification_batch_size)

print(
    f"{len(topics_and_subtopics[0]['content'])} topic(s), "
    f"{len(prompt_questions)} question(s), {len(batches)} batch(es) of up to {settings.classification_batch_size}"
)
```

```python
# Calling classify_batch directly (not through the production
# _classify_batch_safe wrapper) so a bad batch raises here instead of
# silently coming back empty -- easier to debug in a notebook.
batch_results = await asyncio.gather(
    *(classify_batch(topics_and_subtopics, batch) for batch in batches)
)
batch_results
```

```python
resolved: dict[str, list[dict]] = {}
for llm_results in batch_results:
    resolved.update(
        resolve_question_mappings(llm_results, topic_lookup, subtopic_lookup, doc_id="notebook-test")
    )

missed = set(id_map) - set(resolved)
if missed:
    print(f"LLM skipped {len(missed)} of {len(id_map)} question(s): {sorted(missed)}")

classified_questions = [
    {"question": pq["question"], "topic": resolved.get(pq["question_id"], [])}
    for pq in prompt_questions
]
classified_questions
```
